```python


print(f"Versão do PyMuPDF: {pymupdf.__version__}")

# 1. Listar todos os PDFs ordenados pelo nome
pdf_files = sorted(RAW_DATA_DIR.glob("*.pdf"))
print(f"Total de arquivos encontrados: {len(pdf_files)}\n")

# 2. Inspecionar nome e quantidade de páginas de cada arquivo
for pdf_path in pdf_files:
    with pymupdf.open(pdf_path) as doc:
        print(f"📄 {doc.page_count:3d} págs | {pdf_path.name}")
```

```python
# selecionando o primeiro pdf
sample_pdf = pdf_files[0]
print(f"Extraindo amostra de: {sample_pdf.name}\n")
```

```python
# selecionando o primeiro pdf
sample_pdf = pdf_files[0]
print(f"Extraindo amostra de: {sample_pdf.name}\n")

# No PyMuPDF 2.1, chamamos .to_markdown() direto no objeto doc
with pymupdf.open(sample_pdf) as doc:
    pages_data = doc.to_markdown(
        pages=[0, 1, 2, 3, 4],
        page_chunks=True,
        header=False,  # Novo no 2.1: ignora cabeçalhos de página
        footer=False,  # Novo no 2.1: ignora rodapés/numeração de página
    )

# Vamos ver quais chaves o PyMuPDF 2.1 devolve para uma página:
print("Chaves disponíveis em cada página:", list(pages_data[0].keys()))
print("Metadados da página 1:", pages_data[0]["metadata"])

# Visualizar o Markdown gerado nas páginas 1 a 5
for page in pages_data:
    # No PyMuPDF 2.1, a chave mudou de ["page"] para ["page_number"]
    page_num = page["metadata"]["page_number"]
    print(f"\n{'='*40} PÁGINA {page_num} {'='*40}\n")
    print(page["text"][:1500])  # Mostra os primeiros 1500 caracteres de cada página
```
