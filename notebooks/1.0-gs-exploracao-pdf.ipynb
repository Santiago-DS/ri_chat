```python
def extract_data_for_page(pages, path_arq):
    with pymupdf.open(path_arq) as doc:
        pages_data = doc.to_markdown(
            pages=pages,
            page_chunks=True,
            header=False,  # ignora cabecalho
            footer=False,  # ignora rodape
            use_ocr=False
        )
    return pages_data
```

```python
# teste final

def extract_metadata_from_cover(cover_md: str) -> dict:
    """
    Extrai ano, trimestre, período (ex: '1T24') e tipo de demonstração
    a partir do Markdown da primeira página (capa) do relatório.
    """
    # padrao para lidar com '1º', '1°', '4 º' e '1<sup>o</sup>'
    padrao_periodo = r"([1-4])\s*(?:º|°|<sup>o</sup>)?\s*trimestre\s+de\s+(20\d{2})"
    match = re.search(padrao_periodo, cover_md, re.IGNORECASE)

    # teste de match
    if match:
        trimestre = int(match.group(1))
        ano = int(match.group(2))
    else:
        trimestre, ano = None, None

    # Identifica tipo do documento
    if re.search(r"condensad[ao]s", cover_md, re.IGNORECASE):
        tipo_doc = "condensadas"
    elif re.search(r"complet[ao]s", cover_md, re.IGNORECASE):
        tipo_doc = "completas"
    else:
        tipo_doc = "desconhecido"


    result = {
        "trimestre": trimestre,
        "ano": ano,
        "tipo_doc": tipo_doc
    }

    return result
```

# 3.Limpeza dos Dados

Durante  a exploração dos dados, percebi 3 problemas na extração dos textos:
* Páginas vazias (ou quase vazias): algumas páginas estão vindo apenas com quebras de linhas (\n\n\...);
* Tags HTML residuais das tabelas: notei a presença de tags como br, sup, entre outras dentro dos textos;
* Excesso de quebras de linhas: vários blocos têm 5 ou mais quebras de linhas \n\n\n\n\n.

Além desses problemas, existem outros que são tradicionalmente encontrados nesse tipo de problema:

1. **Inconsistência de Encoding e Caracteres Especiais (*Encoding / Mojibake*)**
   - *Problema:* Espaços não-quebráveis (`\xa0`), acentos decompostos ou caracteres corrompidos.
   - *Solução:* Normalização Unicode (**NFKC**).

2. **Hifenização de Quebra de Linha (*Dehyphenation*)**
   - *Problema:* Palavras cortadas no final da linha pela justificação do PDF (ex: `ope-\nracional`).
   - *Solução:* Junção por Regex (`ope-` + `racional` $\rightarrow$ `operacional`).

3. **Tags HTML Residuais em Tabelas (*HTML Artifacts*)**
   - *Problema:* Conversões que deixam tags de formatação visual (ex: `<br>`, `<sup>`, `<sub>`, `<b>`).
   - *Solução:* Substituir `<br>` por espaço e remover tags de sobrescrito para não poluir os tokens.

4. **Ícones e Glifos Invisíveis (*Unicode Private Use Area - PUA*)**
   - *Problema:* Logotipos, setas e marcadores de tópicos do PDF viram caracteres invisíveis (`\ue000`–`\uf8ff`).
   - *Solução:* Remoção via Regex da faixa Unicode privada.

5. **Excesso de Espaços e Quebras em Branco (*Whitespace Pollution*)**
   - *Problema:* Múltiplos espaços horizontais e blocos com mais de 3 quebras de linha (`\n\n\n\n`), roubando espaço útil dos chunks.
   - *Solução:* Reduzir espaços repetidos e limitar quebras a no máximo `\n\n` (parágrafo Markdown).

6. **Páginas Vazias ou Não-Semânticas (*Noise Filtering*)**
   - *Problema:* Capas intermediárias, divisórias de seção e páginas de assinaturas quase sem texto útil.
   - *Solução:* Filtro por tamanho mínimo de caracteres (ex: `< 80` caracteres).

```python
def clean_markdown_page(text: str) -> str:
    """
    Pipeline universal de limpeza para páginas extraídas de PDFs:
    - Normalização Unicode NFKC (corrige \xa0 e acentuações)
    - Desfaz hifenização de fim de linha (ex: ope-\nracional -> operacional)
    - Limpa tags HTML residuais (<br>, <sup>, etc.)
    - Remove caracteres da área privada Unicode (ícones de PDF)
    - Normaliza espaços em branco e excesso de quebras de linha
    """
    if not text:
        return ""
    # 1. Normalização Unicode (converte \xa0 em espaço comum, etc.)
    text = unicodedata.normalize("NFKC", text)

    # 2. Desfaz hifenização de final de linha em português
    text = re.sub(r"([a-zA-ZÀ-ÿ]+)-\n\s*([a-zA-ZÀ-ÿ]+)", r"\1\2", text)

    # 3. Limpeza de HTML residual das tabelas
    text = re.sub(r"<br\s*/?>", " ", text, flags=re.IGNORECASE)

    text = re.sub(r"</?(?:sup|sub|b|i|strong|em|span|small)>", "", text, flags=re.IGNORECASE)

    # 4. Remove caracteres Unicode de área privada (ícones/glifos não mapeados)
    text = re.sub(r"[\ue000-\uf8ff]", "", text)

    # 5. Normaliza espaços horizontais (mantendo quebras de linha)
    text = re.sub(r"[ \t]+", " ", text)

    # 6. Normaliza quebras de linha (no máximo duas quebras: padrão parágrafo Markdown)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()
```

```python
# Testando nas 5 primeiras páginas do primeiro PDF que já extraímos:
sample_pages = extract_data_for_page([0, 1, 2, 3, 4], pdf_files[0])

for page in sample_pages:
    p_num = page["metadata"]["page_number"]
    cleaned = clean_markdown_page(page["text"])
    
    # Vamos considerar página "útil" apenas se tiver pelo menos 50 caracteres após limpeza
    status = "MANTER" if len(cleaned) >= 50 else "DESCARTAR (vazia/curta)"
    print(f"Página {p_num:2d} | {len(cleaned):4d} caracteres | {status}")
```
